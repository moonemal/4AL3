## Comparing with the Baseline

Importing all the model csv files to compare metrics

```python
import pandas as pd

# Load results
baseline = pd.read_csv("baseline.csv")
square_root = pd.read_csv("square_root_weighting.csv")
intermediate = pd.read_csv("intermediate_weighting.csv")
full_weighted = pd.read_csv("full_weighting.csv")

# Standardize the target column
baseline = baseline.rename(columns={"Target": "Complication"})

# Add model names
baseline["Model"] = "Majority Baseline"
square_root["Model"] = "Square-root (0.5)"
intermediate["Model"] = "Intermediate (0.7)"
full_weighted["Model"] = "Full Weighting (1.0)"

# Combine results
comparison = pd.concat(
    [baseline, square_root, intermediate, full_weighted],
    ignore_index=True
)

# Keep the relevant columns
comparison = comparison[
    [
        "Model", "Complication",
        "Accuracy", "Macro Precision",
        "Macro Recall", "Macro F1"
    ]
]

display(comparison.round(4))
```

Comparing the overall performance of the binary complication

```python
binary_comparison = comparison[
    comparison["Complication"] != "LET_IS"
]

binary_overall = (
    binary_comparison
    .groupby("Model", sort=False)[
        ["Accuracy", "Macro Precision", "Macro Recall", "Macro F1"]
    ]
    .mean()
    .reset_index()
)

display(binary_overall.round(4))
```

Comparing the LET_IS 

```python
let_comparison = comparison[
    comparison["Complication"] == "LET_IS"
]

display(let_comparison.round(4))
```
